# This will DELETE all tables in a lakehouse
To use - attach a default lakehouse to the notebook via the left hand menu, then run

```python
# This approach uses the Spark context which often has different permissions than SQL endpoint
from pyspark.sql import SparkSession

# Get a list of all schemas
schemas = spark.sql("SHOW SCHEMAS").collect()

# Loop through each schema
for schema in schemas:
    schema_name = schema.namespace.split('.')[-1].strip("`")
    
    # Skip system schemas
    if schema_name in ['sys', 'information_schema']:
        continue
        
    print(f"\nProcessing schema: {schema_name}")
    
    # Get all tables in the current schema
    tables = spark.sql(f"SHOW TABLES IN {schema_name}").collect()
    
    # Print tables to verify what we'll be deleting
    for table in tables:
        table_name = table.tableName
        print(f"Table found: {schema_name}.{table_name}")
    
    # Delete each table in the current schema
    for table in tables:
        table_name = table.tableName
        try:
            spark.sql(f"DROP TABLE IF EXISTS {schema_name}.{table_name}")
            print(f"Successfully dropped table: {schema_name}.{table_name}")
        except Exception as e:
            print(f"Failed to drop {schema_name}.{table_name}: {str(e)}")
```
